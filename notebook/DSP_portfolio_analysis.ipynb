# DSP Portfolio Analysis
This notebook demonstrates generating a sine wave, adding AWGN, and comparing filtering approaches (moving average and FIR designed with scipy).

```python
from dsp_portfolio_improved import generate_signal, inject_gaussian_noise, apply_causal_lpf_numpy, apply_causal_lpf_streaming, calculate_snr
import matplotlib.pyplot as plt
import numpy as np

FS = 100.0
clean = generate_signal(5, FS, 2.0)
noisy = inject_gaussian_noise(clean, 0.35, seed=42)
filtered = apply_causal_lpf_numpy(noisy, window_size=5)
t = np.arange(len(clean)) / FS
plt.figure(figsize=(10,4))
plt.plot(t, clean, label='clean')
plt.plot(t, noisy, label='noisy', alpha=0.6)
plt.plot(t, filtered, label='filtered')
plt.legend()
plt.show()
```
